```python
import pandas as pd
import numpy as np
```

```python
from pathlib import Path

file_path = Path("../Data/cleaned_loan_dataset.csv")
```

```python
df = pd.read_csv(file_path)
```

```python
if 'Loan_ID' in df.columns:
  df.drop('Loan_ID', axis=1, inplace=True)
```

```python
initial_shape = df.shape
df.drop_duplicates(inplace=True)
print(f"Dropped {initial_shape[0] - df.shape[0]} duplicate rows.")
```

```python
cat_cols_with_na = ['Gender', 'Married', 'Dependents', 'Self_Employed']
for col in cat_cols_with_na:
  if col in df.columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

num_cols_with_na = ['LoanAmount', 'Loan_Amount_Term', 'Credit_History']
for col in num_cols_with_na:
  if col in df.columns:
    df[col].fillna(df[col].median(), inplace=True)

print("Remaining missing values:\n", df.isnull().sum()[df.isnull().sum() > 0])
```

```python
object_cols = df.select_dtypes(include=['object']).columns
for col in object_cols:
  df[col] = df[col].astype(str).str.strip()

if 'Dependents' in df.columns:
  df['Dependents'] = df['Dependents'].replace('3+', '3')

if 'Credit_History' in df.columns:
  df['Credit_History'] = df['Credit_History'].astype(float).astype(int).astype(str)

print("\nData Types after standardizing:\n", df.dtypes)
```

```python
cleaned_file_path = '/content/drive/MyDrive/skill lab project - loan approval intelligence and risk assessment system/cleaned_loan_dataset.csv'
df.to_csv(cleaned_file_path, index=False)

print(f"Cleaned dataset saved successfully to: {cleaned_file_path}")
```
